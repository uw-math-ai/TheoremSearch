```python
import pandas as pd
import re

df = pd.read_csv("math_ag_sample.csv")

def sentence_count(s):
    return len(re.split(r'[.!?]', s))

df["sentence_count"] = df["body"].apply(sentence_count)

df.groupby("manifold")["sentence_count"].agg(["mean", "std"])
```

```python
def inline_math_count(s):
    return s.count("$") - 2 * s.count("$$")

def display_math_count(s):
    return s.count("\\[") + s.count("$$")

df["inline_math"] = df["body"].apply(inline_math_count)
df["display_math"] = df["body"].apply(display_math_count)
df.groupby("manifold")[["inline_math", "display_math"]].mean()
```

```python
def starts_with_math(s):
    return int(s.strip().startswith("$") or s.strip().startswith("\\"))

df["starts_with_math"] = df["body"].apply(starts_with_math)

df.groupby("manifold")["starts_with_math"].mean()
```

```python
LOGIC_TOKENS = ["forall", "exists", "if", "then", "such that", "\\forall", "\\exists"]

def logic_density(s):
    s = s.lower()
    return sum(tok in s for tok in LOGIC_TOKENS)

df["logic_density"] = df["body"].apply(logic_density)

df.groupby("manifold")["logic_density"].mean()
```
